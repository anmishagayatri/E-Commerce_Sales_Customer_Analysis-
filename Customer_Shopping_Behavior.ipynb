```python
from sqlalchemy import create_engine, text
from sqlalchemy.engine import URL

username = "root"
password = "SQL@Learn2026!"
host = "localhost"
port = 3306

url = URL.create(
    "mysql+pymysql",
    username=username,
    password=password,
    host=host,
    port=port
)

engine = create_engine(url)

with engine.connect() as conn:
    conn.execute(text("CREATE DATABASE IF NOT EXISTS customer_behavior"))

print("Database created successfully!")
```

SQLAlchemy manages the database connection/interface, while PyMySQL is the MySQL driver that actually communicates with MySQL.

Python → SQLAlchemy → PyMySQL → MySQL; commit() is used for INSERT/UPDATE/DELETE, while CREATE DATABASE is implicitly committed by MySQL.

SQLAlchemy is a Python library that lets Python communicate with databases like MySQL.

```python
database = "customer_behavior"

url = URL.create(
    "mysql+pymysql",
    username=username,
    password=password,
    host=host,
    port=port,
    database=database
)

engine = create_engine(url)

print("Connected to customer_behavior!")
```

```python
df.to_sql(
    "customer",
    engine,
    if_exists="replace",
    index=False
)
```

df.to_sql() uploads the Pandas DataFrame df into a SQL table called customer using the engine connection.
if_exists="replace" replaces the table if it already exists, and index=False prevents Pandas from adding its index as a column.
Here 0, 1, 2 are the Pandas index

```python
pd.read_sql(
    "SELECT * FROM customer LIMIT 5",
    engine
)
```

pd.read_sql() brings SQL table data back into a Pandas DataFrame so you can analyze/manipulate it in Python.
Pandas df → SQL table: use df.to_sql()
SQL table → Pandas df: use pd.read_sql()
